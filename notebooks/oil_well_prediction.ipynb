# Bootstraping

1000 iteraciones con reemplazo para estimar intervalos de confianza y cuantificar riesgo de pérdida.

```python
lista = [df0_resp, df1_resp, df2_resp]
for df, num in zip(lista, range(len(lista))):
    profits=[]
    df = pd.DataFrame(df)[['predictions', 'target']]
    for i in range(1000):
        sample = df.sample(n=500, replace=True, random_state=54321+i)
        best_sample = sample.sort_values('predictions', ascending=False).head(total_wells)
        profit = fp.calculate_profit(best_sample, total_budge, unit_income, total_wells)
        profits.append(profit)
        profit_mean = best_sample['target'].mean()
        #print(f"Ganancia total (millones de dólares) para la muestra {i+1}: ${profit/10**6:.2f}")
    profits=pd.Series(profits)
    profits_positives = profits[profits > 0].count()
    profits_negatives = profits[profits < 0].count()
    risk= profits_negatives / len(profits)
    profit_lower = profits.quantile(.025)
    profit_upper = profits.quantile(.975)
    print(f"Región {num}")
    print(f"Ganancias positivas: {profits_positives}")
    print(f"Ganancias negativas: {profits_negatives}")
    print(f"Riesgo de pérdidas: {risk*100:.2f}%")
    print(f"Intervalo de confianza del 95% para la ganancia: ${profit_lower/10**6:.2f} - ${profit_upper/10**6:.2f}")
```
